```python
n=int(input("enter a number:\n"))
if n%2==0:
    print(n,"the number is even")
else:
    print(n,"the number is odd")
```

```python
even_sum,odd_sum=0,0
n=int(input("enter a number :\n"))
for i in range(1,n+1):
    if i%2==0:
       even_sum+=i
odd_sum+=i
print("sum of all even numbers:",even_sum)
print("sum of all odd numbers:",odd_sum)
```
